```python
##Program 23: Largest and Smallest in a List

n=5
temp=None
n_list=[10,25,50,5,75]
for a in range(0,n):
    for b in range(a+1,n):
        if n_list[a]>n_list[b]:
            temp=n_list[a]
            n_list[a]=n_list[b]
            n_list[b]=temp
print("Smallest:",n_list[0])
print("Largest:",n_list[n-1])
    
```
